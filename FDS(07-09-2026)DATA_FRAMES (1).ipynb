```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("database.sqlite")

df1 = pd.read_sql("SELECT * FROM Player", conn)
df2 = pd.read_sql("SELECT * FROM Match", conn)
df3 = pd.read_sql("SELECT * FROM Team", conn)
df4 = pd.read_sql("SELECT * FROM Country", conn)

display(df1.head())
display(df2.tail())
display(df3.describe())
display(df4.columns)

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("sql-murder-mystery.db")

df1 = pd.read_sql("SELECT * FROM person", conn)
df2 = pd.read_sql("SELECT * FROM interview", conn)
df3 = pd.read_sql("SELECT * FROM crime_scene_report", conn)
df4 = pd.read_sql("SELECT * FROM income", conn)

display(df1.head())
display(df2.tail())
display(df3.describe(include="all"))
display(df4.columns)

conn.close()
```
